```python
import numpy as np
import matplotlib.pyplot as plt

days=np.array(
    ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']
)
sales=np.array([120,150,135,180,200,250,230])

plt.figure(figsize=(9,5))
plt.plot(
    days,
    sales,
    color="blue",
    linestyle="-",
    linewidth=2,
    label= "Daily Sales",
    marker="o"
    )
plt.title("Weekly Sales Analysis")
plt.xlabel("Days")
plt.ylabel("Sales")

plt.legend()
plt.grid(True)
for i in range(len(days)):
    plt.text(days[i],sales[i]+5, str(sales[i]),ha='center')
plt.show()
```

```python
a=np.array([120,150,135,180,200,250,230])
b=np.array([100,130,160,170,190,210,260])

plt.figure(figsize=(9,5))

plt.plot(
    days,
    a, marker='o',
    color='blue',
    label='Product A'  
        )
plt.plot(
    days,
    b, marker='s',
    color='red',
    label="Product B"
        )
plt.xlabel("Days")
plt.ylabel("Sales")
plt.legend()
plt.grid(True,alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(9,5))

plt.bar(days,sales)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid(axis='y',alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(9,6))

bar=plt.bar(
    days,
    sales,
    color="skyblue",
    edgecolor="black",
    width=0.6
)
plt.bar_label(bar,padding=3)
plt.title("Weekly Sales Analysis",fontsize=16)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid(axis='y',alpha=0.3)
plt.show()
```

```python
x=np.arange(len(days))

plt.figure(figsize=(10,6))
width=0.35
plt.bar(x-width/2,a,width,label='Product A')
plt.bar(x+width/2,b,width,label='Product B')
plt.xticks(x,days)
plt.title("Daily sales : Product A vs Product B",fontsize=16)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.legend()
plt.grid(axis='y',alpha=0.3)
plt.show()
```

```python
x=np.arange(len(days))

plt.figure(figsize=(10,6))
width=0.35
plt.barh(x-width/2,a,width,label='Product A')
plt.barh(x+width/2,b,width,label='Product B')
plt.yticks(x,days)
plt.title("Daily sales : Product A vs Product B",fontsize=16)
plt.ylabel("Days")
plt.xlabel("Sales")
plt.legend()
plt.grid(axis='x',alpha=0.3)
plt.show()
```

```python
sorted_indices=np.argsort(sales)[::-1]
s_days=days[sorted_indices]
s_sales=sales[sorted_indices]

plt.figure(figsize=(9,5))
bar=plt.barh(
    s_days,s_sales,color='teal',edgecolor='black'
)
plt.bar_label(bar,padding=3)
plt.title("Weekly Sales Ranking")
plt.xlabel("Sales")
plt.ylabel("Days")
plt.gca().invert_yaxis()
plt.grid(axis='x',alpha=0.3)
plt.show()
```

```python
marks=np.array([
    45,50,55,60,62,65,67,70,72,74,75,78,80,82,85,88,90,92,95,98
    ])
plt.figure(figsize=(9,5))
plt.hist(marks,bins=10,color='skyblue',density=True,edgecolor='black')
plt.title("Student Marks Distribution")
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.grid(axis='y',alpha=0.3)
plt.show()
```

```python
marks=np.array([
    45,50,55,60,62,65,67,70,72,74,75,78,80,82,85,88,90,92,95,98
    ])
plt.figure(figsize=(9,5))
plt.hist(marks,bins=10,color='skyblue',cumulative=True,edgecolor='black')
plt.title("Cumulative Marks Distribution")
plt.xlabel('Marks')
plt.ylabel('Cumulative Students')

plt.show()
```

```python
hours=np.array([1,2,3,4,5,6,7,8])
scores=np.array([35,40,50,55,65,70,80,90])

plt.figure(figsize=(8,5))
plt.scatter(hours,scores,color='purple',marker='o',s=100,alpha=0.7,edgecolor='black')
plt.title("Study Hours vs Exam Scores")
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.grid(alpha=0.3)
plt.show()
```

```python
m,b=np.polyfit(hours,scores,1)
plt.scatter(hours,scores,s=80)
plt.plot(hours,m*hours+b,color='red')
plt.title("Study Hours vs Exam Scores")
plt.xlabel("Study Hours")
plt.ylabel("Scores")
plt.grid(alpha=0.3)
plt.show()
```

```python
labels=['Python','Java','C++','JavaScript']
students=[40,25,20,15]
plt.figure(figsize=(7,7))
plt.pie(students,labels=labels,autopct='%1.1f%%',startangle=90)
plt.title("Programming Language Preference")
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(students,labels=labels,autopct='%1.1f%%',startangle=90,explode=[0.1,0,0,0],shadow=True)
plt.title("Programming Language Preferencee")
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(students,labels=labels,autopct='%1.1f%%',startangle=90,pctdistance=0.8,labeldistance=1.1,wedgeprops={'width':0.4})
plt.title('Programming Language Preference')
plt.show()
```

```python
marks=np.array([
    45,50,55,60,62,65,67,70,72,74,75,78,80,82,85,88,90,92,95,98,150
])
plt.figure(figsize=(8,5))
plt.boxplot(marks)
plt.title("Student Marks Distribution")
plt.ylabel("Marks")
plt.grid(axis='y',alpha=0.3)
plt.show
```

```python
q1=np.percentile(marks,25)
q2=np.percentile(marks,50)
q3=np.percentile(marks,75)
iqr=q3-q1

print(f"{q1}, {q2}, {q3}, {iqr}")
```

```python
lower=q1-1.5*iqr
upper=q1+1.5*iqr
outlier=marks[(marks<lower) | (marks>upper)]
print(f'{lower}, {upper}, {outlier}')
```

```python
months=np.array(['Jan','Feb','Mar','Apr','May','Jun'])
sales=np.array([100,150,130,180,220,250])
plt.figure(figsize=(9,5))
plt.fill_between(months,sales,alpha=0.4)
plt.plot(months,sales,marker='o')
plt.xlabel("Month")
plt.ylabel('Sales')
plt.grid(alpha=0.3)
plt.show()
```

```python
a=[100,120,140,160,180,200]
b=[80,90,100,110,120,130]
plt.figure(figsize=(9,5))
plt.stackplot(months,a,b,labels=['Product A','Product B'],alpha=0.6)
plt.title("Monthly Combined Product Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend(loc='upper left')
plt.show()
```

```python
fig,ax=plt.subplots(1,2,figsize=(12,5))
ax[0].plot(months,sales,color='blue',marker='o')
ax[0].set_title("Monthly Sales")
ax[0].set_xlabel("Month")

ax[1].bar(months,sales,color='teal')
ax[1].set_title("Monthly Sales Comparison")
ax[1].set_xlabel("Month")

plt.tight_layout()
plt.show()
```

```python
fig,ax=plt.subplots(2,3,figsize=(12,5),sharex=True,sharey=True)

ax[0,0].plot(months,sales)
ax[0,0].set_title("plot")

ax[0,1].bar(months,sales)
ax[0,1].set_title("bar")

ax[0,2].plot(months,sales)
ax[0,2].set_title("plot")

ax[1,0].scatter(months,sales)
ax[1,0].set_title("scatter")

ax[1,1].fill_between(months,sales)
ax[1,1].set_title("area")

ax[1,2].scatter(months,sales)
ax[1,2].set_title("scatter")

plt.tight_layout()
plt.show()
```

```python
hours=np.array([1,2,3,4,5,6,7,8])
scores=np.array([35,40,50,55,65,70,80,90])
plt.figure(figsize=(9,5))
m,b=np.polyfit(hours,scores,1)
plt.plot(hours,m*hours+b,color='red')
plt.scatter(hours,scores,marker='o',s=100)
plt.title("Hours vs Scores Analysis")
plt.xlabel("Hours")
plt.ylabel("Scores")
plt.show()
```
