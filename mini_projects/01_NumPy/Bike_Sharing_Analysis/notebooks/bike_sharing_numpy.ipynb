## Demand Classification Using NumPy

```python
#classify daily rental demand

demand_catagory=np.where(
    rentals<3000,
    "Low",
    np.where(
        rentals<=5000,
        "Medium",
        "High"
    )
)
catagories,counts=np.unique(
    demand_catagory,
    return_counts=True
)
print("DAILY RENTAL DEMAND CLASSIFICATION")
print("----------------------------------")
for catagories,count in zip(catagories,counts):
    percentage=(count/len(rentals))*100

    print(f"{catagories} demand: {count} days ({percentage:.2f}%)")
```

## Seasonal Bike Rental Demand Analysis

```python
# Seasonal demand analysis

seasonal_names={
    1:"Spring",
    2:"Summer",
    3:"Fall",
    4:"Winter"
}
print("SEASONAL BIKE RENTAL ANALYSIS")
print("=" *45)
for season in np.unique(seasons):
    season_rentals=rentals[seasons==season]
    print(f"\nSeason : {seasonal_names[season]}")
    print("Average of daily rentals : ",np.round(np.mean(season_rentals),2))
    print("Median rentals           : ",np.median(season_rentals))
    print("Maximum rentals          : ",np.max(season_rentals))
    print("Minimum rentals          : ",np.min(season_rentals))
    print("Standard deviation       : ",np.round(np.std(season_rentals),2))
```

## Extract weather conditions

```python
with open(file_path,"r") as file:
    reader=csv.DictReader(file)
    data=list(reader)
weather=np.array([int(row['weathersit']) for row in data])

weather_names={
    1:"Clear / Few Clouds",
    2:"Mist / Cloudy",
    3:"Light Rain / Snow",
    4:"Heavy Rain / Snow"
}

print('WEATHER IMPACT ANALYSIS')
print("=" *40)

for condition in np.unique(weather):
    weather_rentals=rentals[condition==weather]
    print(f"\nWeather: {weather_names[condition]}")

    print("Number of days : ",len(weather_rentals))
    print("Average rentals: ",np.round(np.mean(weather_rentals),2))
    print("Total rentals  : ",np.sum(weather_rentals))
    print("Maximum rentals: ",np.max(weather_rentals))
```

## Working Day VS Non-Working Day Analysis

```python
#Extract working-day information

workingday = np.array([int(row['workingday']) for row in data])

day_types={
    0:"Non-Working day",
    1:"Working day"
}

print("WORKING DAY ANALYSIS")
print("=" *40)
for day in np.unique(workingday):
    day_rentals=rentals[workingday==day]
    print(f"\n{day_types[day]}")

    print("Number of days    : ",len(day_rentals))
    print("Total rentals     : ",np.sum(day_rentals))
    print("Average of rentals: ",round(np.mean(day_rentals),2))
    print("Median of rentals : ",np.median(day_rentals))
    print("Maximum of rentals: ",np.max(day_rentals))
```

## Registered VS Casual Customer Analysis

```python
casual=np.array([int(row['casual']) for row in data])
registered=np.array([int(row['registered']) for row in data])
print("CUSTOMER SEGMENT ANALYSIS")
print("=" *40)

print("\nCASUAL CUSTOMERS")
print("Total rentals         : ",np.sum(casual))
print("Average daily rentals : ",round(np.mean(casual),2))
print("Maximum daily rentals : ",np.max(casual))

print("\nREGISTERED CUSTOMERS")
print("Total rental          : ",np.sum(registered))
print("Average daily rentals : ",round(np.mean(registered),2))
print("Maximum daily rentals : ",np.max(registered))

total=np.sum(casual)+np.sum(registered)

print("\nCUSTOMER CONTRIBUTION")
print("Casual    : ",round(np.sum(casual)/total * 100,2),"%")
print("Registered: ",round(np.sum(registered)/total * 100,2),"%")
```
