## 9. Correlation Analysis

Correlation analysis is performed to examine the direction and strength of linear relationships among selected numerical variables.

A correlation heatmap is used to visualize these relationships.

Special attention is given to the correlations between the final grade (`G3`) and other academic and lifestyle variables.

**Note:** Correlation does not imply causation.

```python
correlation_data=df[
    ['studytime','failures','absences','Dalc','Walc','freetime','goout','health','G1','G2','G3']
]
final_corr=correlation_data.corr()
print(final_corr['G3'].sort_values(ascending=False).round(3))
```
