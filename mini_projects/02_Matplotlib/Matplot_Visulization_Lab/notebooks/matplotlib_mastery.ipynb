```python
import numpy as np
import matplotlib.pyplot as plt

days=np.array(
    ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']
)
sales=np.array([120,150,135,180,200,250,230])

plt.figure(figsize=(9,5))
plt.plot(
    days,
    sales,
    color="blue",
    linestyle="-",
    linewidth=2,
    label= "Daily Sales",
    marker="o"
    )
plt.title("Weekly Sales Analysis")
plt.xlabel("Days")
plt.ylabel("Sales")

plt.legend()
plt.grid(True)
for i in range(len(days)):
    plt.text(days[i],sales[i]+5, str(sales[i]),ha='center')
plt.show()
```

```python
a=np.array([120,150,135,180,200,250,230])
b=np.array([100,130,160,170,190,210,260])

plt.figure(figsize=(9,5))

plt.plot(
    days,
    a, marker='o',
    color='blue',
    label='Product A'  
        )
plt.plot(
    days,
    b, marker='s',
    color='red',
    label="Product B"
        )
plt.xlabel("Days")
plt.ylabel("Sales")
plt.legend()
plt.grid(True,alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(9,5))

plt.bar(days,sales)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid(axis='y',alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(9,6))

bar=plt.bar(
    days,
    sales,
    color="skyblue",
    edgecolor="black",
    width=0.6
)
plt.bar_label(bar,padding=3)
plt.title("Weekly Sales Analysis",fontsize=16)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid(axis='y',alpha=0.3)
plt.show()
```

```python
x=np.arange(len(days))

plt.figure(figsize=(10,6))
width=0.35
plt.bar(x-width/2,a,width,label='Product A')
plt.bar(x+width/2,b,width,label='Product B')
plt.xticks(x,days)
plt.title("Daily sales : Product A vs Product B",fontsize=16)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.legend()
plt.grid(axis='y',alpha=0.3)
plt.show()
```

```python
x=np.arange(len(days))

plt.figure(figsize=(10,6))
width=0.35
plt.barh(x-width/2,a,width,label='Product A')
plt.barh(x+width/2,b,width,label='Product B')
plt.yticks(x,days)
plt.title("Daily sales : Product A vs Product B",fontsize=16)
plt.ylabel("Days")
plt.xlabel("Sales")
plt.legend()
plt.grid(axis='x',alpha=0.3)
plt.show()
```

```python
sorted_indices=np.argsort(sales)[::-1]
s_days=days[sorted_indices]
s_sales=sales[sorted_indices]

plt.figure(figsize=(9,5))
bar=plt.barh(
    s_days,s_sales,color='teal',edgecolor='black'
)
plt.bar_label(bar,padding=3)
plt.title("Weekly Sales Ranking")
plt.xlabel("Sales")
plt.ylabel("Days")
plt.gca().invert_yaxis()
plt.grid(axis='x',alpha=0.3)
plt.show()
```
